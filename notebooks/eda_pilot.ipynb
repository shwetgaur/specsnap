Explores source distribution, app categories, OCR coverage, brightness/blur filters.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path('..')
manifest = ROOT / 'data' / 'manifest.csv'
df = pd.read_csv(manifest) if manifest.exists() else pd.DataFrame()
df
```

```python
if not df.empty:
    fig, axes = plt.subplots(1, 2, figsize=(10, 4))
    df['source'].value_counts().plot(kind='bar', ax=axes[0], title='By Source')
    df['app_category'].value_counts().plot(kind='bar', ax=axes[1], title='By Category')
    plt.tight_layout()
    plt.savefig(ROOT / 'results' / 'eda_source_category.png', dpi=120)
    plt.show()
```
